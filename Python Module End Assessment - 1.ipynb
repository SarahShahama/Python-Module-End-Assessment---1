```python
class Bankacc:
  def __init__(self, acc_no, pin_no, balance_amount):
    self.account_number = acc_no
    self.pin_number = pin_no
    self.__balance_amount = balance_amount

  def acc(self):
    x = int(input("Enter your Account Number:"))
    y = int(input("Enter your Pin Number:"))
    if x == self.account_number and y == self.pin_number:
      print("Login Successful")
      z = int(input("Enter your Deposit Amount:"))
      if z>0:
        print("Deposit Successful")
        self.__balance_amount += z
        print("Your  Balance is", self.__balance_amount)
      else:
        print("Invalid Deposit Amount")
    else:
      print("Invalid Account Number or PIN")

  def withdraw(self):
    x = int(input("Enter your Account Number:"))
    y = int(input("Enter your pin Number:"))
    if x == self.account_number and y == self.pin_number:
      print("Login Successful")
      g = int(input("Enter your Withdrawal Amount:"))
      if g > 0 and g <= self.__balance_amount:
        print("Withdrawal Successful")
        self.__balance_amount -= g
        print("Your Balance is", self.__balance_amount)
      else:
        print("Invalid Withdrawal Amount or Insufficient Balance")
    else:
      print("Invalid Account Number or PIN")


person1 = Bankacc(123456789, 1234, 100)
person1.acc()
print("------------------------------------------------------")
person2 = Bankacc(234567891, 4321, 100)
person2.acc()
print("------------------------------------------------------")
person3 = Bankacc(453672837, 9900, 500)
person1.acc()
print("------------------------------------------------------")
person4 = Bankacc(342512627, 4531, 800)
person4.withdraw()
print("------------------------------------------------------")
person5 = Bankacc(563829726, 5642, 400)
person5.withdraw()
print("------------------------------------------------------")
person6 = Bankacc(349283628, 4982, 300)
person6.withdraw()

```
